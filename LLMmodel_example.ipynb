```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler as scale
```

```python
data= pd.read_excel('/content/drive/MyDrive/STATISTICS-TEST/Numerical/AirQualityUCI.xlsx')
data.head(10)
```

```python
# Dropping Date and Time column to convert dataset into numerical format instead of Time-Series
data.drop(['Date','Time'],axis=1,inplace=True)
```

```python
X=data.drop(['T'],axis=1)
Y=data['T']
```

```python
x_train,x_test,y_train,y_test=train_test_split(data,Y,test_size=0.15)
```

```python
from LMsyn import LMmodel

model = LMmodel( batch_size=8, epochs=1)
model.fit(x_train)
```

```python
# model.save("/content/drive/MyDrive/lm_syn")
```

```python
path="/content/drive/MyDrive/lm_syn_furth"
```

```python
import os
import torch
import json
import numpy as np
def save( path):
        if not os.path.isdir(path):
            os.mkdir(path)

        with open(path + "/config.json", "w") as f:
            attributes = model.__dict__.copy()
            attributes.pop("tokenizer")
            attributes.pop("model")
            if isinstance(attributes["cond_col_dist"], np.ndarray):
                attributes["cond_col_dist"] = list(attributes["cond_col_dist"])
            json.dump(attributes, f)
        torch.save(model.model.state_dict(), path + "/model.pt")
save(path)
```

```python
from sdmetrics.reports.single_table import QualityReport

report = QualityReport()
report.generate(x_test, samples, metadata)
```
